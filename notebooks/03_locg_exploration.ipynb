```python
from bs4 import BeautifulSoup
from pathlib import Path

html = Path("absolute.html").read_text(
    encoding="utf-8"
)

soup = BeautifulSoup(html, "html.parser")

print(soup.title.get_text())
```

```python
links = soup.find_all("a", href=True)

for link in links:
    if "/comic/" in link["href"]:
        print(link["href"], link.get_text(strip=True))
```

```python
variant_links = soup.select('a[href*="?variant="]')

print("Total variant links:", len(variant_links))
```

```python
print(variant_links[0].prettify())
```

```python
from urllib.parse import urlparse, parse_qs, urljoin

BASE_URL = "https://leagueofcomicgeeks.com"

variant = variant_links[0]

href = variant["href"]
parsed_url = urlparse(href)
query_params = parse_qs(parsed_url.query)

image = variant.find("img")

record = {
    "issue_id": int(parsed_url.path.split("/")[2]),
    "variant_id": int(query_params["variant"][0]),
    "variant_name": variant["title"].strip(),
    "full_title": image.get("alt"),
    "image_url": image.get("data-src"),
    "variant_url": urljoin(BASE_URL, href)
}

record
```

```python
import requests

url = "https://leagueofcomicgeeks.com/comic/1669159/absolute-batman-23"

headers = {
    "User-Agent": "AbsolutePipeline/0.1 (Personal research project)",
    "Accept": "text/html,application/xhtml+xml",
    "Accept-Language": "en-US,en;q=0.9",
}

response = requests.get(
    url,
    headers=headers,
    timeout=30
)

print(response.status_code)
```

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(response.text, "html.parser")

print("Title:", soup.title.get_text(strip=True))

variant_links = soup.select('a[href*="?variant="]')

print("Total variants:", len(variant_links))
```

```python
print(variant_links[0].prettify())
```

```python
from urllib.parse import urlparse, parse_qs

variant_ids = [
    int(parse_qs(urlparse(link["href"]).query)["variant"][0])
    for link in variant_links
]

print("Total variants:", len(variant_ids))
print("Unique variants:", len(set(variant_ids)))
```

```python
images = [
    link.find("img")
    for link in variant_links
]

image_urls = [
    image.get("data-src")
    for image in images
    if image is not None and image.get("data-src")
]

print("Total image URLs:", len(image_urls))
print("Unique image URLs:", len(set(image_urls)))
```
